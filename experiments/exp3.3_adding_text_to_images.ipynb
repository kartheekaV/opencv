```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread(r"doremon.png")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

img_text = img.copy()

cv2.putText(img_text, "Hello OpenCV", (50,50),
            cv2.FONT_HERSHEY_SIMPLEX,
            1, (255,0,0), 2)

plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
plt.imshow(img)
plt.title("Original Image")

plt.subplot(1,2,2)
plt.imshow(img_text)
plt.title("Image with Text")

plt.show()
```
