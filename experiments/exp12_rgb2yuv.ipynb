```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("shinchan.jpg")

rgb_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

yuv_img = cv2.cvtColor(img, cv2.COLOR_BGR2YUV)

plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
plt.imshow(rgb_img)
plt.title("Original RGB")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(yuv_img)
plt.title("YUV Image")
plt.axis('off')

plt.show()
```
