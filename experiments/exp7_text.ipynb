```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("shinchan.jpg")

cv2.putText(
    img,
    "Hello Shinchan",
    (50, 100),
    cv2.FONT_HERSHEY_SIMPLEX,
    2,
    (255, 0, 0),
    3
)

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.axis('off')
plt.show()
```
