```python
import cv2
import matplotlib.pyplot as plt

# Read image
img = cv2.imread("shinchan.jpg")

# Convert BGR to RGB
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Flip horizontally
flip = cv2.flip(img, 1)

# Display image
plt.imshow(flip)
plt.title("Flipped Image")
plt.axis('off')
plt.show()
```
